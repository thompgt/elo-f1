# F1 Driver Elo Standings by Season

One cell per season (1980–present). Each cell queries `driver_elo_season_summary` for that year and shows the drivers ordered by end-of-season Elo rating, alongside their championship points and finishing position.

Run `python -m elo_f1.elo.run_elo` first if `data/elo_f1.db` is stale.

```python
import sqlite3
import pandas as pd

DB_PATH = "data/elo_f1.db"


def elo_standings(year: int) -> pd.DataFrame:
    conn = sqlite3.connect(DB_PATH)
    try:
        df = pd.read_sql_query(
            """
            SELECT
                d.given_name || ' ' || d.family_name AS driver,
                c.name AS constructor,
                s.elo_season_end,
                s.elo_season_average,
                s.points,
                s.final_points_position,
                s.races_started
            FROM driver_elo_season_summary s
            JOIN drivers d ON d.driver_id = s.driver_id
            JOIN constructors c ON c.constructor_id = s.constructor_id
            WHERE s.year = ?
            ORDER BY s.elo_season_end DESC
            """,
            conn,
            params=(year,),
        )
    finally:
        conn.close()
    df.index = range(1, len(df) + 1)
    df.index.name = "elo_rank"
    return df
```
